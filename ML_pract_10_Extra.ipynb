```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import StandardScaler

# Step 2: Upload CSV file
from google.colab import drive
drive.mount("/content/drive")

path= "/content/drive/MyDrive/ML_Dataset/Spotify_Songs_Dataset.csv"
df=pd.read_csv(path)
df.head(5)

print("First 5 rows:")
print(df.head())

print("\nColumns:")
print(df.columns)
```

```python
# Select only numeric columns
X = df.select_dtypes(include=['float64', 'int64'])

# Optional: drop non-useful columns like id or index
X = X.drop(columns=[col for col in X.columns if 'id' in col.lower()], errors='ignore')

print("Selected Features:")
print(X.head())
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Step 5: Create Dendogram
plt.figure(figsize=(8,5))

dendrogram(linkage(X_scaled,method='ward'))

plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean Distance')
plt.show()
```

```python
hc = AgglomerativeClustering(n_clusters=5, metric='euclidean', linkage='ward')

clusters = hc.fit_predict(X_scaled)

# Add cluster labels to dataset
df['Cluster'] = clusters

print(df.head())
```
