```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import pandas as pd

# Re-run train_test_split to ensure X_train and X_test are DataFrames
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Identify categorical columns in X_train
categorical_cols = X_train.select_dtypes(include='object').columns

# Apply one-hot encoding to both X_train and X_test
X_train_processed = pd.get_dummies(X_train, columns=categorical_cols, drop_first=True)
X_test_processed = pd.get_dummies(X_test, columns=categorical_cols, drop_first=True)

# Align columns - ensuring both train and test sets have the same columns
# This handles cases where a category might appear in train but not test, or vice-versa.
all_columns = X_train_processed.columns.union(X_test_processed.columns)
X_train_processed = X_train_processed.reindex(columns=all_columns, fill_value=0)
X_test_processed = X_test_processed.reindex(columns=all_columns, fill_value=0)

# Ensure the order of columns is the same for both, crucial for consistent scaling
X_test_processed = X_test_processed[X_train_processed.columns]

# Initialize and apply StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train_processed)
X_test = scaler.transform(X_test_processed)
```

**Vector Machine**

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import numpy as np

# Define a threshold for classification (e.g., exam_score > 30 is 'Pass', else 'Fail')
# You can adjust this threshold based on your domain knowledge or problem requirements
threshold = 30
y_train_classified = (y_train > threshold).astype(int) # Convert to 0 or 1
y_test_classified = (y_test > threshold).astype(int)   # Convert to 0 or 1

svc=SVC()
svc.fit(X_train,y_train_classified)
y_pred=svc.predict(X_test)
accuracy=accuracy_score(y_test_classified,y_pred)
print('Model accuracy score with default hyperparameters: {0:0.4f}'. format(accuracy))
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
print('Confusion Matrix')
cm = confusion_matrix(y_pred,y_test_classified)
print(confusion_matrix(y_pred,y_test_classified))
```

```python
print('Accuracy Score')
print(accuracy_score(y_pred,y_test_classified))
```

```python
print('Classification Report')
print(classification_report(y_pred,y_test_classified))
```
